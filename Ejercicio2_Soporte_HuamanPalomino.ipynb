## Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Tiempo de espera, nivel de urgencia y condición de cliente premium. |
| **Acciones (A)** | Escalar el ticket a nivel 1, nivel 2 o nivel 3. |
| **Entorno (E)** | Mesa de ayuda, clientes, tickets y niveles de soporte técnico. |
| **Objetivo** | Asignar cada ticket al nivel adecuado, priorizando casos urgentes y evitando esperas excesivas. |
| **Medida de desempeño (P)** | Tiempo de resolución, cantidad de tickets atendidos oportunamente y satisfacción del cliente. |

### Justificación de las reglas

Las reglas buscan equilibrar la urgencia del problema, el tiempo de espera y la condición
del cliente. Los tickets de urgencia alta se escalan inmediatamente para reducir el tiempo
de respuesta ante problemas críticos. Los clientes premium pueden recibir atención prioritaria,
mientras que los tickets de baja urgencia se escalan cuando el tiempo de espera aumenta.

```python
def agente_soporte(
    tiempo_espera_minutos,
    nivel_urgencia,
    cliente_premium
):
    nivel_urgencia = nivel_urgencia.lower()

    if nivel_urgencia == "alta":
        return (
            "escalar a nivel 3",
            "urgencia alta reportada por el cliente"
        )

    elif nivel_urgencia == "media":

        if tiempo_espera_minutos >= 30:
            return (
                "escalar a nivel 3",
                "urgencia media con espera de 30 minutos o más"
            )

        elif cliente_premium:
            return (
                "escalar a nivel 3",
                "cliente premium con urgencia media"
            )

        else:
            return (
                "escalar a nivel 2",
                "urgencia media sin condiciones adicionales"
            )

    elif nivel_urgencia == "baja":

        if tiempo_espera_minutos >= 60:
            return (
                "escalar a nivel 2",
                "urgencia baja con espera de 60 minutos o más"
            )

        elif cliente_premium and tiempo_espera_minutos >= 30:
            return (
                "escalar a nivel 2",
                "cliente premium con espera elevada"
            )

        else:
            return (
                "escalar a nivel 1",
                "urgencia baja y espera dentro del límite"
            )

    else:
        return (
            "revisar manualmente",
            "nivel de urgencia no válido"
        )
```

### Simulación y pruebas

```python
casos_soporte = [
    (5, "alta", False),
    (10, "media", False),
    (35, "media", False),
    (10, "media", True),
    (70, "baja", False),
    (40, "baja", True)
]

for i, caso in enumerate(casos_soporte, start=1):

    tiempo, urgencia, premium = caso

    accion, motivo = agente_soporte(
        tiempo,
        urgencia,
        premium
    )

    print(f"Caso {i}")
    print(f"Tiempo de espera: {tiempo} minutos")
    print(f"Nivel de urgencia: {urgencia}")
    print(f"Cliente premium: {premium}")
    print(f"Acción: {accion}")
    print(f"Motivo: {motivo}")
    print("-" * 60)
```
