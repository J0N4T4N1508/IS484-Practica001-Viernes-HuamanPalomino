## Inspección visual de un producto

### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Matriz NumPy de 5x5 donde 0 representa un píxel normal y 1 un píxel defectuoso. |
| **Acciones (A)** | Aprobar, enviar a revisión manual o rechazar el producto. |
| **Entorno (E)** | Línea de producción industrial con productos y cámara de inspección. |
| **Objetivo** | Detectar productos defectuosos y tomar una decisión automática. |
| **Medida de desempeño (P)** | Precisión en la detección, reducción de productos defectuosos aprobados y tiempo de inspección. |

### Justificación de las reglas

Las reglas permiten separar los productos según la cantidad de defectos detectados.
Un producto sin defectos puede aprobarse directamente, mientras que una cantidad pequeña
de defectos requiere revisión manual. Cuando la cantidad de defectos supera el límite establecido,
el producto se rechaza para evitar que un producto con demasiadas anomalías continúe en producción.

```python
import numpy as np

def agente_inspeccion(imagen):

    defectos = np.sum(imagen == 1)

    if defectos == 0:
        return (
            "aprobar",
            "no se detectaron pixeles defectuosos"
        )

    elif defectos <= 3:
        return (
            "revision manual",
            f"se detectaron {defectos} pixeles defectuosos"
        )

    else:
        return (
            "rechazar",
            f"se detectaron {defectos} pixeles defectuosos"
        )
```
